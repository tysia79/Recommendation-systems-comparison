```python
import numpy as np 
import pandas as pd
import os
```

```python
data='/Users/Tysia/Desktop/Edukacja/Informatyka/Magisterka/ml-1m'
rating_data=os.path.join(data,'ratings.dat')
movie_data=os.path.join(data,'movies.dat')
user_data=os.path.join(data,'users.dat')

ratings=pd.read_csv(
  rating_data,
  sep='::',
  engine='python',
  names=['user','item','rating','timestamp'],
  encoding='latin-1'   
)

movies=pd.read_csv(
    movie_data, 
    sep='::',
    engine='python',
    names=['item', 'title', 'genres'],
    encoding='latin-1'
)
users = pd.read_csv(
    user_data, 
    sep='::',
    engine='python',
    names=['user', 'gender', 'age', 'occupation', 'zipcode'],
    encoding='latin-1'
)
```

```python
print(ratings.head())
print(movies.head())
print(users.head())
```

```python
import subprocess
from sklearn.model_selection import train_test_split
```

```python
path_to_mml = "C:\\Users\\Tysia\\Desktop\\Edukacja\\Informatyka\\Magisterka\\Mymedialite\\libraries\\MyMediaLite-3.11\\lib\\mymedialite\\rating_prediction.exe"

# pliki tymczasowe
train_file = 'train.data'
test_file = 'test.data'

df_clean = ratings[['user', 'item', 'rating']].copy()

# probelmy z kropkami i przecinkami, więc rzutuję na int dla ocen 1-5
df_clean['rating'] = df_clean['rating'].astype(int)
df_clean['user'] = df_clean['user'].astype(int)
df_clean['item'] = df_clean['item'].astype(int)
```

```python
train, test = train_test_split(df_clean, test_size=0.2, random_state=42)

# zapisywanie danych do pliku tymczasowego
train.to_csv(train_file, sep='\t', header=False, index=False)
test.to_csv(test_file, sep='\t', header=False, index=False)
```

```python
# budowanie komendy ręcznie
#i-knn
cmd_str = (
    f'"{path_to_mml}" '           
    f'--training-file="{train_file}" '
    f'--test-file="{test_file}" '
    f'--recommender=ItemKNN '
    f'--recommender-options k=20 '
#    f'--online-evaluation'
)
```

```python
#u-knn
cmd_str = (
    f'"{path_to_mml}" '           
    f'--training-file="{train_file}" '
    f'--test-file="{test_file}" '
    f'--recommender=UserKNN '
    f'--recommender-options k=20 '
#    f'--online-evaluation'
)
```

```python
#mostpop
cmd_str = (
    f'"{path_to_mml}" '           
    f'--training-file="{train_file}" '
    f'--test-file="{test_file}" '
    f'--recommender=MostPopular '
)
```

```python
# matrix factorisation
cmd_str = (
    f'"{path_to_mml}" '
    f'--training-file="{train_file}" '
    f'--test-file="{test_file}" '
    f'--recommender=MatrixFactorization' 
)
```
